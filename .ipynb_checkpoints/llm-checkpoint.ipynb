```python
from bpe import BPETokenizer
from dataloader import Dataset, DataLoader
```

```python
import urllib.request
import os
url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
file_path = "dataset.txt"

if not os.path.exists(file_path):
    urllib.request.urlretrieve(url, file_path)

with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()
```

```python
TARGET_VOCAB = 512
tokenizer = BPETokenizer()
tokenizer.train(text, TARGET_VOCAB)
```

```python
encoding = tokenizer.encode("Hello, my name is Szymon")
print(encoding)
decoding = tokenizer.decode(encoding)
print(decoding)
```

```python
dataset = Dataset(tokenizer.encode(text), 100, 15, TARGET_VOCAB)
dataloader = DataLoader(dataset, 2, True)
for inp, tar in dataloder:
    print(inp)
    print(tar)
```
